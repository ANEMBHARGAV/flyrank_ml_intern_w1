## 4. The unit of analysis, as a real dataframe

*Load your lane's slice and show it: one row = one what?*

### Unit of analysis

The unit of analysis is one page. Each row in the dataframe represents one anonymized content page. The features describe the page's observed search performance and content characteristics, while the target represents whether the page is labeled as declining.

```python
import os
import subprocess
import pandas as pd

REPO_URL = "https://github.com/ANEMBHARGAV/flyrank_ml_intern_w1"
REPO_DIR = "flyrank_ml_intern_w1"

if not os.path.exists(REPO_DIR):
    subprocess.run(["git", "clone", "--depth", "1", REPO_URL, REPO_DIR], check=True)

df = pd.read_csv(
    os.path.join(REPO_DIR, "data", "raw", "content_refresh_anonymized.csv")
)

# Create the target from the defined starter rule
df["is_declining_label"] = (
    df["trend_direction"].str.lower() == "down"
).astype(int)

# Select observable features and the target
lane_df = df[
    [
        "impressions_90d",
        "ctr",
        "avg_position",
        "content_age_days",
        "days_since_last_update",
        "is_declining_label"
    ]
].copy()

print("One row = one anonymized content page")
print("Rows:", len(lane_df))
print("Columns:", list(lane_df.columns))

lane_df.head()
```

## 5. Why ML beats a fixed rule here

*What makes the pattern too messy for an if-statement?*

### Why ML beats a fixed rule here

A fixed rule could flag pages using one simple condition, such as content age or a drop in impressions. However, page performance can depend on several signals at the same time, and a single threshold may miss useful combinations of signals. A classification model can learn patterns from multiple observed features and provide a ranked review list for manual checking. The model should still be treated as decision-support rather than an automatic decision.
